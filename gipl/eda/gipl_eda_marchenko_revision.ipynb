`````python
def meta_check(extract_dir):

    all_meta = []
    
    fps = [x for x in extract_dir.glob("*.tif")]

    read_lock = threading.Lock()

    def process(fp):
        src = rasterio.open(fp)
        with read_lock:
            profile = src.profile
            if profile not in all_meta:
                all_meta.append(profile)
    
    # We map the process() function over the list of files
    with concurrent.futures.ThreadPoolExecutor(
        max_workers=32
    ) as executor:
        executor.map(process, fps)

    return all_meta
`````

The next step in our exploration process is to visualize and generate a few summary statistics of a sample of the dataset.

`````python
# initialize a dictionary that will sample all variables
crrel_di = {}

permafrost_var_names = ["Mean Annual Ground Temperature 0.5 m", "Mean Annual Ground Temperature 1 m",
                        "Mean Annual Ground Temperature 2 m", "Mean Annual Ground Temperature 3 m",
                        "Mean Annual Ground Temperature 4 m", "Mean Annual Ground Temperature 5m",
                        "Permafrost Base", "Permafrost Top", "Mean Annual Surface Temperature", "Talik"]

permafrost_var_file_tags = ["magt_0.5m", "magt_1m", "magt_2m", "magt_3m", "magt_4m", "magt_5m",
                             "pfBase", "pfTop", "Surf", "Talik"]

permafrost_var_units = ["°C", "°C", "°C", "°C", "°C", "°C", "m", "m", "°C", "m"]

permafrost_var_cmaps = ["magma", "magma", "magma", "magma", "magma", "magma", "cool", "winter", "magma", "viridis"]


for pf_var, tag, unit, cmap in zip(permafrost_var_names, permafrost_var_file_tags,
                                   permafrost_var_units, permafrost_var_cmaps):
    crrel_di[pf_var] = {}
    crrel_di[pf_var]["file tag"] = tag
    crrel_di[pf_var]["unit"] = unit
    crrel_di[pf_var]["cmap"] = cmap
`````

`````python
def get_nine_random_geotiffs(permafrost_var):
    glob_str = "*{0}*".format(permafrost_var)
    fps = [x for x in crrel_extract_dst.glob(glob_str)]
    nine_fps = random.sample(fps, 9)
    return nine_fps

def iem_get_nine_random_geotiffs(permafrost_var):
    glob_str = "*{0}*".format(permafrost_var)
    fps = [x for x in iem_extract_dst.glob(glob_str)]
    nine_fps = random.sample(fps, 9)
    return nine_fps

def get_year(fp):
    year = re.match(r'.*([1-3][0-9]{3})', fp).group(1)
    return year


def get_model(fp):
    model = re.match(r".*(ccsm4|cm3|cru|5modelavg)", fp.lower()).group(1)
    return model


def get_scenario(fp):
    
    try:
        scenario = re.match(r".*(rcp[0-9][0-9])", fp.lower()).group(1)
    except:
        scenario = "historical"
    return scenario


def get_stat_summary_from_raster_list(rstrs, file_tag):
    
    di = {}
    
    for fp in rstrs:
        trimmed_fp = fp.name[:-4]
        year = get_year(trimmed_fp)
        model = get_model(trimmed_fp)
        scenario = get_scenario(trimmed_fp)
        fp_key = year + " " + model + " " + scenario
        di[fp_key] = {}
        with rasterio.open(fp) as src:
            arr = src.read(1)
            arr[arr == src.nodata] = np.nan
            di[fp_key]["fp"] = fp
            di[fp_key]["arr"] = arr
            di[fp_key]["min"] = np.round(np.nanmin(arr), 2)
            di[fp_key]["max"] = np.round(np.nanmax(arr), 2)
            di[fp_key]["mean"] = np.round(np.nanmean(arr), 2)
            di[fp_key]["sd"] = np.round(np.nanstd(arr), 2)
            di[fp_key]["width"] = arr.shape[1]
            di[fp_key]["height"] = arr.shape[0]
            di[fp_key]["year"] = year
            di[fp_key]["model"] = model
            di[fp_key]["scenario"] = scenario
            di[fp_key]["variable"] = file_tag
            di[fp_key]["pixel_size_m"] = src.profile["transform"][0]
    return di
`````

`````python
for k in crrel_di.keys():
    nine_fps = get_nine_random_geotiffs(crrel_di[k]["file tag"])
    crrel_di[k]["sample"] = get_stat_summary_from_raster_list(nine_fps, crrel_di[k]["file tag"])
`````

`````python
def get_all_max_of_di(di):
    max_vals = []
    for k in di.keys():
        max_vals.append(di[k]["max"])
    return max(max_vals)

    
def get_all_min_of_di(di):
    min_vals = []
    for k in di.keys():
        min_vals.append(di[k]["min"])
    return min(min_vals)


def df_from_di(di):
    df = pd.DataFrame.from_dict(di).T
    del df["arr"]
    return df


def plot_maps(di, title, unit, cmap, stretch=1.0, region="ak"):

    fig, axs = plt.subplots(3, 3, figsize=(12, 12))
    fig.patch.set_facecolor("white")
    plt.suptitle(title + " (" + unit + ")", y=0.9)
    
    vmin = get_all_min_of_di(di)
    vmax = get_all_max_of_di(di)
    
    vmax_stretch = vmax * stretch
    vmin_stretch = vmin * stretch
    
    for ax, k in zip(axs.flatten(), di.keys()):
        arr = di[k]["arr"]
        if vmin == 0:
            cvmin = vmax * 0.01
            cvmax = vmax * 0.20
            arr[arr == 0] = np.nan
        else:
            cvmin = vmin_stretch
            cvmax = vmax_stretch
        ax.imshow(arr, vmin=cvmin, vmax=cvmax, cmap=cmap, interpolation="none")
        ax.set_title(k)

    cbar_ax = fig.add_axes([0.15, 0.05, 0.7, 0.06])
    fig.colorbar(axs[0,0].images[0], cax=cbar_ax, orientation="horizontal", extend="both")

    #plt.show()
    plt.savefig(title + "_" + region + ".png", dpi=144, bbox_inches="tight")
`````

`````python
for k in crrel_di.keys():
    plot_maps(crrel_di[k]["sample"], k, crrel_di[k]["unit"], crrel_di[k]["cmap"], 0.9)
`````

`````python
sample_dfs = []
for k in crrel_di.keys():
    crrel_di[k]["sample"]["df"] = df_from_di(crrel_di[k]["sample"])
    sample_dfs.append(crrel_di[k]["sample"]["df"])

summary_df = pd.concat(sample_dfs)
summary_df
`````
